```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding , LSTM , Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
reviews = [
    'i like the movie',
    'the film was amazing',
    'i hate the movie',
    'the file is terrible',
    'i really enjoyed the movie',
    'the movie is not good',
    'movie is boring',
    'film maker was bastard'
]
labels = [1,1,0,0,1,0,0,
0]
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(reviews)
sequences = tokenizer.texts_to_sequences(reviews)
```

```python
print('Sequences : ',sequences)
```

```python
# padding is must
X = pad_sequences(sequences)
y = np.array(labels)
```

```python
vocab_size = len(tokenizer.word_index) + 1
max_len = X.shape[1]
```

```python
model = Sequential([
    #Input Layer - embedding layer(we are passing data in bits)
    Embedding(input_dim = vocab_size , output_dim =10,  mask_zero=True), # sir used input_len
    LSTM(units =32),
    Dense(units =1 ,activation = "sigmoid") # here units is one so sigmoid ,else
])
```

```python
model.compile(loss ='binary_crossentropy',optimizer ='adam')
# one hot encoding ni hota to sparse
# here binary answer we are having
# sparse
'''
Adam (Adaptive Moment Estimation) is an advanced optimization algorithm that:
(learning rate ko running time pe optimise krta h automatically)
Combines momentum (past gradients memory)
adaptive learning rate (adjusts step size automatically)
'''
```

```python
model.fit(X,y,epochs=50)
```
