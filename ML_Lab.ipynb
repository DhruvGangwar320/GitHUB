```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('/content/USA_Housing.csv')
```

```python
obj = LinearRegression()
X = df[['Area Population','Avg. Area Income','Avg. Area House Age','Avg. Area Number of Rooms','Avg. Area Number of Bedrooms']]
Y = df['Price']
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(X, Y, test_size = 0.2 , random_state = 42)
obj.fit(x_train, y_train)
pred = obj.predict(x_test)
print(pred)
```

```python
# Mean error
from sklearn import metrics

mae = metrics.mean_absolute_error(y_test, pred)
mse = metrics.mean_squared_error(y_test, pred)
rmse = np.sqrt(mse)

print(f'Mean Absolute Error (MAE): {mae}')
print(f'Mean Squared Error (MSE): {mse}')
print(f'Root Mean Squared Error (RMSE): {rmse}')
```

#Implement K-Nearest Neighbors to classify different types of seeds using the Seeds Dataset.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

df1 = pd.read_csv('/seed_data(1).csv', sep=',')
df= pd.DataFrame(df1)
X = df[['area','perimeter','compactness','kernel_length','kernel_width','asymmetry','groove_length']] # Corrected 'assymetery' to 'asymmetry'
y = df['class']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)


knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_scaled, y_train)

y_pred_knn = knn.predict(X_test_scaled)

print(f'Accuracy (KNN): {accuracy_score(y_test, y_pred_knn):.2f}')
print('\nClassification Report (KNN):')
print(classification_report(y_test, y_pred_knn))
print('\nConfusion Matrix (KNN):')
print(confusion_matrix(y_test, y_pred_knn))
```

#Implement Support Vector Classification (SVC) to predict the presence of heart disease using the Heart Disease Dataset.

Identify and impute missing values in both 'df1' and 'df2'. This will involve using appropriate strategies for numerical (e.g., mean/median) and categorical (e.g., mode/new category) columns.


**Reasoning**:

**Reasoning**:
Based on the identified missing values, I will now impute numerical columns with their median values from `df1` and categorical columns with their mode from `df1`. This ensures consistent imputation across both dataframes and prevents data leakage from the test set.



```python
numerical_cols_with_na_df1 = df1.select_dtypes(include=np.number).columns[df1.select_dtypes(include=np.number).isnull().any()].tolist()
categorical_cols_with_na_df1 = df1.select_dtypes(include='object').columns[df1.select_dtypes(include='object').isnull().any()].tolist()

numerical_cols_with_na_df2 = df2.select_dtypes(include=np.number).columns[df2.select_dtypes(include=np.number).isnull().any()].tolist()
categorical_cols_with_na_df2 = df2.select_dtypes(include='object').columns[df2.select_dtypes(include='object').isnull().any()].tolist()

# Impute numerical columns in df1 and df2 using medians from df1
for col in numerical_cols_with_na_df1:
    if col != 'Id': # Don't impute Id column
        median_val = df1[col].median()
        df1[col].fillna(median_val, inplace=True)
        df2[col].fillna(median_val, inplace=True)

# Impute additional numerical columns in df2 not present in df1's missing list (e.g., BsmtFinSF1)
for col in numerical_cols_with_na_df2:
    if col not in numerical_cols_with_na_df1 and col != 'Id':
        median_val = df1[col].median() if col in df1.columns else 0 # Use 0 if not in df1 (e.g. BsmtFinSF1, BsmtFinSF2 etc.)
        df2[col].fillna(median_val, inplace=True)

# Impute categorical columns in df1 and df2 using modes from df1
for col in categorical_cols_with_na_df1:
    mode_val = df1[col].mode()[0] # .mode() can return multiple values, take the first
    df1[col].fillna(mode_val, inplace=True)
    df2[col].fillna(mode_val, inplace=True)

# Impute additional categorical columns in df2 not present in df1's missing list (e.g., MSZoning)
for col in categorical_cols_with_na_df2:
    if col not in categorical_cols_with_na_df1:
        mode_val = df1[col].mode()[0] if col in df1.columns else 'None'
        df2[col].fillna(mode_val, inplace=True)

print("Missing values have been imputed.")
```

**Reasoning**:
The previous code produced `FutureWarning` due to `inplace=True` with chained assignment in Pandas. To address this, I will modify the code to explicitly assign the result of `fillna` back to the DataFrame column, which is the recommended practice for future Pandas versions.



```python
numerical_cols_with_na_df1 = df1.select_dtypes(include=np.number).columns[df1.select_dtypes(include=np.number).isnull().any()].tolist()
categorical_cols_with_na_df1 = df1.select_dtypes(include='object').columns[df1.select_dtypes(include='object').isnull().any()].tolist()

numerical_cols_with_na_df2 = df2.select_dtypes(include=np.number).columns[df2.select_dtypes(include=np.number).isnull().any()].tolist()
categorical_cols_with_na_df2 = df2.select_dtypes(include='object').columns[df2.select_dtypes(include='object').isnull().any()].tolist()

# Impute numerical columns in df1 and df2 using medians from df1
for col in numerical_cols_with_na_df1:
    if col != 'Id': # Don't impute Id column
        median_val = df1[col].median()
        df1[col] = df1[col].fillna(median_val)
        df2[col] = df2[col].fillna(median_val)

# Impute additional numerical columns in df2 not present in df1's missing list (e.g., BsmtFinSF1)
for col in numerical_cols_with_na_df2:
    if col not in numerical_cols_with_na_df1 and col != 'Id':
        median_val = df1[col].median() if col in df1.columns else 0 # Use 0 if not in df1 (e.g. BsmtFinSF1, BsmtFinSF2 etc.)
        df2[col] = df2[col].fillna(median_val)

# Impute categorical columns in df1 and df2 using modes from df1
for col in categorical_cols_with_na_df1:
    mode_val = df1[col].mode()[0] # .mode() can return multiple values, take the first
    df1[col] = df1[col].fillna(mode_val)
    df2[col] = df2[col].fillna(mode_val)

# Impute additional categorical columns in df2 not present in df1's missing list (e.g., MSZoning)
for col in categorical_cols_with_na_df2:
    if col not in categorical_cols_with_na_df1:
        mode_val = df1[col].mode()[0] if col in df1.columns else 'None'
        df2[col] = df2[col].fillna(mode_val)

print("Missing values have been imputed.")
```
