```python
import math
# import matplotlyb.pyplot as plt
```

```python
w1 , w2 , b1 = 0.15 , 0.2 , 0.1
# Hidden neuron 2
w3 ,w4 ,b2 = 0.25 , 0.3 , 0.10
# output neuron
w5 ,w6 , b3 = 0.4 , 0.45 , 0.10
learning_rate = 0.1
```

```python
x ,x2 = 1, 1
target = 1
```

```python
def sigmoid(x):return 1/(1+math.exp(-1*x))
def der_sigmoid(x):return sigmoid(x)*(1-sigmoid(x))
```

```python
def forward_pass(x1,x2):
  h1net = w1*x1 + b1 + w2*x2
  h1 = sigmoid(h1net)
  h2net = w3*x2 + b2 + w4*x2
  h2 = sigmoid(h2net)
  global ynet
  ynet = w5*h1 + w6*h2 + b3
  output = sigmoid(ynet)
  return h1,h2,output
h1 , h2 , output =forward_pass(x,x2)
print(h1,h2,output)
y = output
print('target == ' ,target)
```

```python
#Loss calculation
loss = 0.5*(target - output)**2
print(loss)
```

#BACKWARD PROOOGATION

```python
# Calculate Gradient

# Gradients for the output layer weights and bias
# dL/dOutput (error from loss function)
dl_doutput = y - target

# dOutput/dYnet (derivative of sigmoid for output)
doutput_dynet = der_sigmoid(ynet)

# dL/dYnet (gradient of loss with respect to the pre-activation of output layer)
dl_dynet = dl_doutput * doutput_dynet

# Gradients for w5, w6, b3
dl_dw5 = dl_dynet * h1
dl_dw6 = dl_dynet * h2
dl_db3 = dl_dynet * 1 # derivative of ynet wrt b3 is 1

# Gradients propagated back to hidden layer outputs (h1, h2)
dl_dh1 = dl_dynet * w5
dl_dh2 = dl_dynet * w6

# Calculate h1net and h2net again for their derivatives
# Note: x is x1, x2 is x2 from the initial cell 527W7UqivcYh
h1net_val = w1*x + b1 + w2*x2
h2net_val = w3*x2 + b2 + w4*x2 # Based on forward_pass definition

# Gradients for hidden layer 1 (h1) weights and bias
dh1_dh1net = der_sigmoid(h1net_val) # Derivative of sigmoid for h1
dl_dh1net = dl_dh1 * dh1_dh1net

dl_dw1 = dl_dh1net * x
dl_dw2 = dl_dh1net * x2
dl_db1 = dl_dh1net * 1 # derivative of h1net wrt b1 is 1

# Gradients for hidden layer 2 (h2) weights and bias
dh2_dh2net = der_sigmoid(h2net_val) # Derivative of sigmoid for h2
dl_dh2net = dl_dh2 * dh2_dh2net

dl_dw3 = dl_dh2net * x2 # derivative of h2net wrt w3 is x2
dl_dw4 = dl_dh2net * x2 # derivative of h2net wrt w4 is x2
dl_db2 = dl_dh2net * 1 # derivative of h2net wrt b2 is 1

# Update weights and biases
global w1, w2, w3, w4, w5, w6, b1, b2, b3, learning_rate

w1 -= learning_rate * dl_dw1
w2 -= learning_rate * dl_dw2
w3 -= learning_rate * dl_dw3
w4 -= learning_rate * dl_dw4
w5 -= learning_rate * dl_dw5
w6 -= learning_rate * dl_dw6

b1 -= learning_rate * dl_db1
b2 -= learning_rate * dl_db2
b3 -= learning_rate * dl_db3

print("Weights and biases updated successfully.")
```
