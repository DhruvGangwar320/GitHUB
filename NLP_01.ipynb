# **NPL PHASE 01 : TEXT PROCESSING**


```python
text = "i am feeling very very happy today !! .."
print("Original text : " , text)
```

```python
# // convert text into lower case
#Step 1 : Convert text into , lowercase
# bucause : "Happy" != "happy"
# ML consider them different

text = text.lower()
print("Lower case text : " , text)
```

```python
# Remove the punctuation .
#Step 2 : we remove everything except lettes and expression
# using re - regular expression
# we will make pattern through with re if the text matches  with it it is
import re
text = re.sub(r'[^\w\s]','',text) #What does empty string does
print("Without punctuation : ",text)
```

```python
# Step - 03 - Tokenization (split sentence into words)
tokens = text.split()
print("Tokens : " ,tokens)
```

```python
# Step - 04 remove the stop words - i , am , is , are , the  , a , an
stopwords = ['i','am','is' ,'are' ,'the','an ','very']
# way 1 : tokens = set(tokens) - set(stopwords)
#way2 :
filtered_tokens = [token for token in tokens if token not in stopwords]
print("After removing the stopwords : " ,filtered_tokens)
```

```python
def stem(word):
  if word.endswith("ing"):
    return word[:-3]
  elif word.endswith("ly") or word.endswith('es') or word.endswith('ed'):
    return word[:-2]
  else:
    return word
stemming  = []
for t in filtered_tokens:
  stemming.append(stem(t))
print("After stemming : " ,stemming)
```

# **NLP PHASE 02 : WORDS TO VECTOR**

```python
#WHY vectors ?
#Because NLP only understands numbers
#So we must convert words into numeric form

#Bag of words
```

```python
# This helps us to convert text -> count
from sklearn.feature_extraction.text import CountVectorizer
```

```python
sentences = ["i am good" , "I am happy" ,"i am okay" ,"happy today"]
```

```python
# Vectorize
vectorizer = CountVectorizer() # Create object
X = vectorizer.fit_transform(sentences)
#^^^ learn vocabulary and convert sentences to numbers
```

```python
print("Vocabulary")
print(vectorizer.get_feature_names_out()) # features are columns which is x data and there is y data which is target feature
```

```python
# Vocabulary
# ['am' 'good' 'happy' 'okay' 'today']  every element is column
```

```python
print("Vectors :")
print(X.toarray())
```

```python
# Here ^^ :
# Each row = sentence
# Each column = word
# Each value = count
```

```python
sentences = [
    'i am learning python and python is good',
    'python is a great programming language.',
    'i also love learn ml with python.'
]
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(sentences)
print("Vocabulary : ",vectorizer.get_feature_names_out())
print("\nMatrix")
print(X.toarray())
```

#**TF-IDF -Term Frequency  -Inverse Document Frequency**


```python
# TF-IDF -Term Frequency  -Inverse Document Frequency
# Bag of words treats all words equally
# For example : I am learning python , python and python
# It will assign huge wait to python
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
sentences = [
    'i am learning python and python is good',
    'python is a great programming language.',
    'i also love learn ml with python.'
]
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(sentences)
# Internally it will :
# Build vocabulary
# Calculate TF
# Calculate IDF
# Multiple tf x idf
# Generate matrix
```

```python
print("Vocabulary : ", vectorizer.get_feature_names_out())
```

```python
print("TF - IDF Matrix : ")
print(X.toarray())
```
