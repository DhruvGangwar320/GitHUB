# INTENT CLASSIFICATION

Vectorization
Vectorization is the process of converting textual data into numerical vectors. Most machine learning algorithms cannot directly process raw text; they require numerical input. Vectorization transforms words or documents into a numerical representation that captures their meaning and relationships.

In this notebook, specifically in cells e0CjH7nPpYbe, GBt9oJQwqHNO, WulU5fywgBip, and 5ot1sC_Ri8P3, you're using TfidfVectorizer for vectorization.

TF-IDF (Term Frequency-Inverse Document Frequency) is a popular vectorization technique. It works by:

Term Frequency (TF): Measuring how frequently a term appears in a document. The more often a word appears, the higher its TF.
Inverse Document Frequency (IDF): Measuring how important a term is across the entire corpus. Words that appear frequently in many documents (like 'the', 'a', 'is') get a lower IDF score, reducing their weight. Rare words get a higher IDF.
By multiplying TF and IDF, TF-IDF gives more weight to terms that are unique and important to a specific document, while downplaying common words. The output is a matrix where each row represents a document (or sentence, in your case) and each column represents a unique word (or N-gram), with the cell values being the TF-IDF scores.

Masking
Masking is a concept predominantly used in deep learning and neural networks, especially with sequence models (like Recurrent Neural Networks or Transformers), but it's not directly applied in the TF-IDF and Logistic Regression setup in this notebook.

In general NLP, masking is used to:

Handle variable sequence lengths: When processing sentences of different lengths, shorter sentences are often 'padded' with special tokens to match the length of the longest sentence. Masking ensures that these padding tokens don't influence the model's calculations (e.g., during attention mechanisms or loss computation).
Attend to specific parts of a sequence: In tasks like machine translation or text generation, masking can prevent a model from 'looking ahead' at future tokens it hasn't generated yet, or restrict its attention to specific parts of the input.
Pre-training tasks: In models like BERT, masking involves randomly hiding (masking) some words in a sentence and then training the model to predict those masked words, helping it learn contextual representations of language.
Since your current notebook uses TfidfVectorizer and LogisticRegression, which are classical machine learning techniques, explicit 'masking' as a distinct pre-processing step is not utilized here. The function cell performs text cleaning (lowercasing, tokenization, punctuation removal, stopword removal, lemmatization), which is a common pre-processing step before vectorization.

```python
# First import text processing libraries
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

#Second import sklearn dependencies
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
```

```python
# stemming , lemitization
# 4 major uses of nltk
# a > Tokenization
# b > Stopword removal
# c > stemming
# d >  Lemmatization
nltk.download("punkt_tab") # Sentence tokenization , word tokenization
nltk.download("punkt")
nltk.download("stopwords") # Downloads a list of common words
nltk.download("wordnet") # Downloads the WordNet lexical database Lemmatization
# ^^^ Understanding word meaning & relationships

```

```python
# (Problem , intent) classification
training_data = [
    ("hello" ,'greet'),
    ("hi" ,'greet'),
    ("how are you" ,'greet'),
    ("good morning",'greet'),
    ("good afternoon",'greet'),
    ("good evening",'greet'),

     ("bye" ,'farewell'),
    ("goodbye" ,'farewell'),
    ("see you later" ,'farewell'),

    ('what is the weather today','weather'),
    ('tell me weather','weather'),
    ('is it raining','weather'),
    ('what is the temperature','weather'),
    ('how hot is it','weather'),
    ('how cold is it','weather'),

    ('open','open_app'),
    ('launch','open_app'),
    ('run','open_app'),
    ('execute','open_app'),
    ('start','open_app'),
    ('open google','open_app'),
    ('open youtube','open_app'),
    ('open gmail','open_app'),
    ('open mail','open_app'),
    ('close','close_app'),
    ('quit','close_app'),
    ('exit','close_app')
]
```

```python
sentences = []
labels = []

for text , intent in training_data :
    sentences.append(text)
    labels.append(intent)
```

```python
import string
# creating function for n - sentences
def function(text_group):
  # convert into lower
  cleaned_sentences = []
  for text in text_group:

    #Lowering text
    text = text.lower()

    # extracting tokens group
    tokens = word_tokenize(text)

    #Removing the punctuation
    punc_filtered  = [word for word in tokens if word not in string.punctuation]

    #Stopword filteration
    stopword_filter = [word for word in punc_filtered if word not in stopwords.words("english")]

    #Word lemmatization
    wnet = WordNetLemmatizer()
    lemmitized_words  = [wnet.lemmatize(word,"v") for word in stopword_filter]

    text = " ".join(lemmitized_words)
    cleaned_sentences.append(text)
  return cleaned_sentences
```

```python
#clean data
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Clean the training sentences using the defined function
cleaned_training_sentences = function(sentences)

# Initialize TfidfVectorizer
vectorizer = TfidfVectorizer()
"""
X = vectorizer.fit_transform(cleaned_training_sentences):

This is a two-in-one step:
fit(cleaned_training_sentences): The vectorizer learns from your cleaned_training_sentences. It identifies all the unique words (your vocabulary) and calculates their Inverse Document Frequencies (IDF) based on how often they appear across all your training sentences.


"""

# Fit and transform the cleaned training sentences
X = vectorizer.fit_transform(cleaned_training_sentences)

print("Shape of TF-IDF matrix:", X.shape)
print("First 5 feature names:", vectorizer.get_feature_names_out()[:5])
```

```python
try:
    # Initialize and train the Logistic Regression model
    logistic = LogisticRegression()
    logistic.fit(X, labels) # X: features  , labels: material which has to be classified .
    print("Logistic Regression model trained successfully.")
except ValueError as e:
    print(f"Error training Logistic Regression model: {e}")
    print("This error usually means that the number of samples in 'X' (features) and 'labels' (targets) do not match.")
    print("Please ensure that 'X' is generated from all training sentences (27 samples) by re-running the preceding cells, specifically 'GU3tAM60nCo1' to populate 'sentences' and 'labels' correctly, and then 'e0CjH7nPpYbe' to generate 'X' from these sentences.")
```

```python
user_input = "love"
processed = function([user_input]) # Wrap user_input in a list
user_input_vector = vectorizer.transform(processed)
prediction = logistic.predict(user_input_vector)
print(prediction)
```

```python
# Purpose: To limit the "vocabulary" size.If you have 10,000 documents, you might end up with 50,000 unique words.
# This can make your model slow and prone to overfitting.
# max_features tells sklearn to only keep the top $N$ most frequent words based on their importance.
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    "The quick brown fox jumps over the lazy dog.",
    "Machine learning is amazing and full of surprises.",
    "Python is the best language for data science."
]

# Configure the vectorizer
vectorizer = TfidfVectorizer(
    stop_words='english',      # Remove common noise
    ngram_range=(1, 2),        # Catch phrases like "machine learning"
    max_features= 4        # Keep the data manageable
)

# Apply to data
X = vectorizer.fit_transform(corpus)

# Look at what we created
print("""Output with vectorizer = TfidfVectorizer(
    stop_words='english',      # Remove common noise
    ngram_range=(1, 2),        # Catch phrases like "machine learning"
    max_features= 1000         # Keep the data manageable
    :>Feature count: 27
['amazing' 'amazing surprises' 'best' 'best language' 'brown' 'brown fox'
 'data' 'data science' 'dog' 'fox']

)
 """)
print(f"Current o/p:>Feature count: {len(vectorizer.get_feature_names_out())}")
print(vectorizer.get_feature_names_out()[:10]) # Show first 10
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction.text import TfidfVectorizer

# Sample Data: Text and Labels (1 = Positive, 0 = Negative)
X_train_raw = ["I love this!", "This is terrible.", "Amazing work!", "Total waste of time."]
y_train = [1, 0, 1, 0]

# 1. Transform text to numbers
vectorizer = TfidfVectorizer()
X_train = vectorizer.fit_transform(X_train_raw)

# 2. Initialize the model
# 'max_iter' is often increased because text data can be "sparse"
model = LogisticRegression(max_iter=1000)

# 3. Train the model
model.fit(X_train, y_train)

# 4. Predict on new data
new_text = ["This is amazing!"]
X_new = vectorizer.transform(new_text)
prediction = model.predict(X_new)

print(f"Prediction: {prediction[0]}") # Output: 1
```
