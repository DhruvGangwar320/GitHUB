```python
import nltk
```

```python
# stemming , lemitization
# 4 major uses of nltk
# a > Tokenization
# b > Stopword removal
# c > stemming
# d >  Lemmatization
nltk.download("punkt_tab") # Sentence tokenization , word tokenization
nltk.download("punkt")
nltk.download("stopwords") # Downloads a list of common words
nltk.download("wordnet") # Downloads the WordNet lexical database Lemmatization
# ^^^ Understanding word meaning & relationships

```

```python
from nltk.tokenize import word_tokenize # word_tokenize
from nltk.corpus import stopwords # for removing stopwords
from nltk.stem import PorterStemmer , WordNetLemmatizer # for lemmatation and stemming
```

```python
text = "Today's day is as beautiful as my thoughts . Since world is the reflection of your brain"
print("Original text : " , text)
```

```python
#Step - 01 lowercase()
text = text.lower()
print("Lowercase text : " , text)
```

```python
# Step - 02 : Tokenize
tokens = word_tokenize(text)
print(tokens)
```

```python
# Step -03
# lets remove punctuation
import string
punc = string.punctuation
print(punc)
```

```python
punctuation_filter = [word for word in tokens if word not in punc]
print("Removed punctuations : ",punctuation_filter)
```

```python
# Step - 04
# Remove stopwords
filtered_tokens = [word for word in punctuation_filter if word not in stopwords.words("english")]
print("Removed stopwords : " , filtered_tokens)
```

```python
# Step - 05 Stemming & Lemmitization
stem = PorterStemmer()
stem.stem("flying")
stem.stem("playing")
```

```python
wnet = WordNetLemmatizer()
print(wnet.lemmatize("playing",'v'))
print(wnet.lemmatize("cooked",'v'))
print(wnet.lemmatize("went",'v'))
print(wnet.lemmatize("bought",'v'))
```

```python
lemmitized_words = [wnet.lemmatize(word,"v") for word in filtered_tokens]
print("Lemmatization : " , lemmitized_words)
```

```python
text = " ".join(lemmitized_words)
text
```

To improve lemmatization, especially for words that can be different parts of speech, we can first perform Part-of-Speech (POS) tagging. This helps the lemmatizer apply the correct rules for each word.

```python
# creating function for n - sentences
def function(text_group):
  # convert into lower
  cleaned_sentences = []
  for text in text_group:
    #Lowering text
    text = text.lower()
    # extracting tokens group
    tokens = word_tokenize(text)
    punc_filtered  = [word for word in tokens if word not in string.punctuation]
    print("Removed Punctuations : " , punc_filtered)
    stopword_filter = [word for word in punc_filtered if word not in stopwords.words("english")]
    print("Stopwords filtered : " , stopword_filter)
    lemmitized_words  = [wnet.lemmatize(word,"v") for word in stopword_filter]
    print("Lemitized word : "  , lemmitized_words)
    text = " ".join(lemmitized_words)
    cleaned_sentences.append(text)
  return cleaned_sentences
sentences  =[
    'i am good',
    'i am playing',
    'i am studying'
]
clean_sentence = function(sentences)
clean_sentence
```
